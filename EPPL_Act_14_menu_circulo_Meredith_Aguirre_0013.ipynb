```python
import math

def areacirculo():
    print("\n--- 🟢 Calculando Área ---")
    radio = float(input("Introduce el valor del radio: "))
    area = math.pi * (radio ** 2)
    print(f"✨ El área del círculo es: {area:.2f} u²")

# Función para calcular el perímetro del círculo
def perimetrocirculo():
    print("\n--- 🔵 Calculando Perímetro ---")
    radio = float(input("Introduce el valor del radio: "))
    perimetro = 2 * math.pi * radio
    print(f"✨ El perímetro es: {perimetro:.2f} u")

# Función para calcular el volumen de la esfera
def volumenesfera():
    print("\n--- 🟣 Calculando Volumen ---")
    radio = float(input("Introduce el valor del radio: "))
    volumen = (4/3) * math.pi * (radio ** 3)
    print(f"✨ El volumen de la esfera es: {volumen:.2f} u³")

# Función principal del menú
def menu():
    continuar = True

    while continuar:
        print("\n" + "="*30)
        print("    🍎 ASISTENTE GEOMÉTRICO 🍎")
        print("="*30)
        print("1️⃣ Calcular área círculo")
        print("2️⃣ Calcular perímetro círculo")
        print("3️⃣ Calcular volumen esfera")
        print("4️⃣ Salir ❌")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == "1":
            areacirculo()
        elif opcion == "2":
            perimetrocirculo()
        elif opcion == "3":
            volumenesfera()
        elif opcion == "4":
            print("\n👋 ¡Gracias por usar el programa! Hasta pronto. ✨")
            continuar = False
        else:
            print("\n⚠️ Opción no válida, intenta de nuevo.")

# Llamada inicial al programa
menu()
```
